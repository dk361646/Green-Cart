```python
import re
import string
```

```python
#cleaning the textual columns to be used
def clean_text(text):
    if isinstance(text, str):
        text = text.lower()
        text = re.sub(r'\d+', '', text)
        text = text.translate(str.maketrans('', '', string.punctuation))
        text = text.strip()
        text = re.sub('\s+', ' ', text)
        return text
    else:
        return ""
df_clean.loc[:, 'title'] = df_clean['title'].apply(clean_text)
df_clean.loc[:, 'description'] = df_clean['description'].apply(clean_text)
df_clean.loc[:, 'material'] = df_clean['material'].apply(clean_text)
df_clean.loc[:, 'brand'] = df_clean['brand'].apply(clean_text)
```

```python
#defining Lemmatization function
import spacy

# Load the English model
nlp = spacy.load("en_core_web_sm")

def lemmatize_text(text):
    txt = nlp(text.lower())
    return " ".join([token.lemma_ for token in txt
                     if not token.is_stop and token.is_alpha])
```

```python
#lemmatization
df_clean.loc[:,'text'] = df_clean.loc[:, 'title'] + " " + df_clean.loc[:,'description']+ " " + df_clean.loc[:,'name']
df_clean['lemmatized_description'] = (df_clean['description'].astype(str)).apply(lemmatize_text)
description = df_clean.loc[:,'lemmatized_description'].astype(str).tolist()
```

```python
def get_sustainability_score(material_text):
    if isinstance(material_text, str):
        material_text = material_text.lower()

        if 'organic' in material_text or 'bamboo' in material_text or 'hemp' in material_text:
            return 1.0
        elif 'recycled' in material_text or 'sustainable' in material_text:
            return 0.8
        elif 'cotton' in material_text or 'linen' in material_text:
            return 0.7
        elif 'polyester' in material_text or 'nylon' in material_text or 'leather' in material_text:
            return 0.3
        else:
            return 0.5  # Neutral if unknown
    else:
        return 0.5  # Neutral for missing material
```

```python
# Add a new column for sustainability score in dataset
df_clean.loc[:,'sustainability_score'] = df_clean.loc[:,'text'].apply(get_sustainability_score)
sus_score = df_clean.loc[:,'sustainability_score']
df_clean.loc[:,['text', 'sustainability_score']].head()
```

```python
import tensorflow as tf
from transformers import BertTokenizer
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
# Tokenization
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
tokens = tokenizer(
    description,
    padding=True,
    truncation=True,
    return_tensors="tf",
    max_length=128
)
```

```python
input_ids = tokens["input_ids"]
attention_mask = tokens["attention_mask"]
input_ids = input_ids.numpy()
attention_mask = attention_mask.numpy()
input_ids.shape, attention_mask.shape
```

```python
from sklearn.model_selection import train_test_split

# Train-test split
X_train_ids, X_test_ids, X_train_mask, X_test_mask, y_train, y_test = train_test_split(
    input_ids,
    attention_mask,
    sus_score,
    test_size=0.2,
    random_state=42
)
```

```python
# building the model
from transformers import TFAutoModel

# Load pretrained BERT
bert_model = TFAutoModel.from_pretrained("bert-base-uncased")

# Enable BERT fine-tuning
bert_model.trainable = True

# Inputs: make sure these are either NumPy arrays or tf.Tensors with shape (num_samples, 128)
# input_ids, attention_mask = ...

batch_size = 8  # You can reduce this if you still get OOM errors
num_samples = input_ids.shape[0]
all_cls_outputs = []

for i in range(0, num_samples, batch_size):
    input_ids_batch = tf.convert_to_tensor(input_ids[i:i+batch_size])
    attention_mask_batch = tf.convert_to_tensor(attention_mask[i:i+batch_size])

    # Run BERT inference
    bert_outputs = bert_model({'input_ids': input_ids_batch, 'attention_mask': attention_mask_batch})
    cls_output = bert_outputs.last_hidden_state[:, 0, :]  # extract CLS token (shape: [batch_size, 768])

    all_cls_outputs.append(cls_output)

# Concatenate all CLS outputs
final_cls_outputs = tf.concat(all_cls_outputs, axis=0)  # shape: (total_samples, 768)

# You can now pass `final_cls_outputs` to your downstream model (like a dense layer)


input_cls = Input(shape=(768,))
x = Dense(64, activation='relu')(input_cls)
x = Dense(32, activation='relu')(x)
output = Dense(1)(x)

model = Model(inputs=input_cls, outputs=output)

#Compile
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=2e-5),
              loss='mse',
              metrics=['mae'])
```

```python
# Train
early_stopping = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
history = model.fit(
    x=final_cls_outputs[0:2869,:],
    y=y_train,
    validation_split=0.1,
    epochs= 10,
    batch_size= 8,
    callbacks=[early_stopping]
)
```

```python
# Evaluate
loss, mae = model.evaluate(final_cls_outputs[2869:,:], y_test)
print(f"Test MAE: {mae:.4f}")
```

```python
# function to recommend similar products
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def recommend_similar_products(product_index, embeddings, n=5):

    selected_embedding = embeddings[product_index].reshape(1, -1)
    selected_category = df_clean.iloc[product_index]['category']

    # Step 2: Filter products by same category
    category_mask = df_clean['category'] == selected_category
    filtered_embeddings = embeddings[category_mask]
    filtered_df = df_clean[category_mask].copy()

    # Step 2: Compute cosine similarity with all other product embeddings
    similarities = cosine_similarity(selected_embedding, filtered_embeddings)[0]
    filtered_df['similarity'] = similarities
    filtered_df = filtered_df.drop(index=df_clean.index[product_index], errors='ignore')
    top_recommendations = filtered_df.sort_values(by='similarity', ascending=False).head(n)

    # Step 3: Get indices of the most similar products
    #similar_indices = np.argsort(similarities)[::-1][1:top_n+1]  # [1:] to skip itself

    # Step 4: Retrieve product information
    #recommended_products = df_clean.iloc[similar_indices][['name', 'title', 'brand', 'material']]

    return top_recommendations[['name', 'title', 'category', 'brand', 'material', 'sustainability_score', 'similarity']]
```

```python
def get_sustainability_info(product):
    material = product['material']  # or description or brand
    explanation = sustainability_explanation(material)

    return {
        'explanation': explanation
    }
```

```python
def recommend_sustainable_products(product_index, embeddings, top_n=5, sustainability_weight=0.5, threshold=0.7):
    """
    Recommend products with sustainability awareness and green badges.
    """

    # Step 1: Check sustainability of selected product
    selected_product = df_clean.iloc[product_index]
    selected_sustainability = selected_product['sustainability_score']
    explaination = sustainability_explanation(selected_product['material'])

    if selected_sustainability >= threshold:
        print(f"The selected product ('{selected_product['name']}') is sustainable! (Score: {selected_sustainability}) (Explaination: {explaination})")
    else:
        print(f"The selected product ('{selected_product['name']}') is NOT very sustainable. (Score: {selected_sustainability}) (Explaination: {explaination})")

    print("\nRecommending more sustainable similar products...\n")

    #getting similar products from the defined function
    similar_products = recommend_similar_products(product_index, final_cls_outputs.numpy(), n = top_n*5)

    #getting sustainable products
    sustain_products = similar_products[similar_products['sustainability_score'] > 0.7]

    return sustain_products.iloc[:top_n, :]
```
